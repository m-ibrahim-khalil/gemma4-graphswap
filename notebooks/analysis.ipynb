# Graph resolution study: analysis

Thin wrapper over `notebooks/analysis.py`, which holds all the logic. Running this notebook rewrites `results/summary/*.csv`, `results/summary/tables.md` and the figures in `paper/figures/` from the run CSVs in `results/runs/` and the audit CSVs in `results/`. The same outputs come from `uv run python notebooks/analysis.py`.

Statistics: `graphswap.stats` (McNemar exact, task-clustered bootstrap); decision rules: `docs/preregistration.md`.

Partial phases (C1 missing a seed, a run at a seed C1 lacks) are noted, skipped and marked `INCOMPLETE` in `tables.md` and in the `incomplete` columns; the final paper regeneration uses `--strict`, which raises on them instead.

```python
import sys
from pathlib import Path

import pandas as pd
from IPython.display import Image, Markdown, display

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'pyproject.toml').exists())
sys.path.insert(0, str(ROOT / 'notebooks'))
import analysis

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
# strict=False: partial phases are noted, skipped and marked INCOMPLETE;
# the final paper regeneration runs `uv run python notebooks/analysis.py --strict`.
out = analysis.run(ROOT / 'results', ROOT / 'paper' / 'figures', strict=False)
```
